```python
# Step 1: Data Collecting - Alreaady we have data preprocessing dataset so no need to do data collection
# Step 2: Data Preprocessing - We have data pre-processed dataset in .csv format which nas been named as 'Salary_Data.csv'
# use pandas library to perform all the database or file or IO activities
# For use pandas library use comments import pandas as pd
# here pd is shortcut or alian of pandas
import pandas as pd
```

```python
# by using pandas - read_csv() function read whoke dataset from the 'Salary_Data.csv'files
# read dataset and save it in dataset variable
dataset = pd.read_csv("Salary_Data.csv");
# execute dataset variable and see the output
# dataset
```

```python
# Step :3 - split input and output is started
# Get input column 'YearsExperience' from save dataset variable by using below code
# Which can be called as independent variable
independent = dataset[["YearsExperience"]]
# execute independent variable and see the output
# independent
```

```python
# Get output column 'Salary' from save dataset variable by using below code
# Which can be called as dependent variable
dependent = dataset[["Salary"]]
# execute dependent variable and see the output
# dependent
```

```python
# Step :4 - split train and test set is started
# use sklearn library help to create model by doing split train and test set, creaate linear regression, find weight and bias
# import train_test_split class from sklearn library model_selection namespace by using below code
# By using this class we can create train and test set
# pass 4 input params - independent,dependent,test_size=0.30,random_state=0
# test_size=0.30 - help to calculate how many percentage can allocate for test and train set ex. here 30 % for test set and 70% for train set
# get 4 output values - x_train,x_test,y_train,y_test as 2 set of x and y train set and 2 set of x and y train set
from sklearn.model_selection import train_test_split
# train_test_split(independent,dependent,test_size=0.30,random_state=0)
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
# Step :5 - Model train and creation is started
# use sklearn library linear_model to simple linear regression (SLR) odel
# import LinearRegression class from sklearn library linear_model namespace by using below code
# after regressor.fit it will create simple linear regression
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(x_train,y_train)
```

```python
# calculate weight from simple linear regression and save it in weight variable by using regressor.coef_
weight = regressor.coef_
```

```python
# calculate bias from simple linear regression and save it in bias variable by using regressor.intercept_
bias = regressor.intercept_
```

```python
# calculate predicted output from simple linear regression and save it y_pred variable by using regressor.predict where passsing input as test set (x_test)
y_pred = regressor.predict(x_test)
```

```python
# Step :6 - Test Set and Evalution Metrics is started
# use sklearn library metrics namespace r2_score class to perform Evalution Metrics which help to validate to created model
# import r2_score class from sklearn library metrics namespace by using below code
from sklearn.metrics import r2_score
# after call r2_score(y_test,y_pred) it will create R2 value and save it r_score variable
# we have to pass inputs of actual output value(y_test) and predicted output(y_pred)
r_score = r2_score(y_test,y_pred)
```

```python
#if see value of R2
# if value near 0 is bad model so model has not been trained well
# if value near 1 is good model so model has been trained well
r_score
```

```python
#6.	Save the Best Model 
# use pickle library which help to save the created model
# import pickle library by using below code
import pickle
filename = "finalized_model_linear.sav"
```

```python
# after call dump(regressor,open(filename,'wb')) it will save the model 
# we have to pass inputs of regressor, save file name with write action 'wb'
pickle.dump(regressor,open(filename,'wb'))
```

```python
# II. Deployment-Phase-2
# 1. Load the Save Model
# Load saved model to the variable load_model by using  pickle.load(open(filename,'rb'))
# here we pass saved file name with action read 'rb'
load_model = pickle.load(open(filename,'rb'))
# 2. Get Inputs - we can pass input dynamicaclly or statically toget predicted output
# 3. Predicts
result = load_model.predict([[12]])
```
