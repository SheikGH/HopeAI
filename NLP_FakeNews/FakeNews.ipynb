```python
import pandas as pd
import numpy as np
import itertools
#Vectors - Counter, TD-IDF, Hasshing
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.model_selection import train_test_split
# New Algorithms - PassiveAggressiveClassifiation, MultinomialNB
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn import metrics
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('news.csv',index_col=None)
# Clean dataset => remove - rows axis=0 (default) or remove columns axis=1
dataset=df.drop('Unnamed: 0',axis=1)
#dataset
```

```python
# Input / Output - x = text, y = label - text = input, label = FAKE/REAL
y=dataset["label"]
#y
```

```python
# give input - independent, output - dependent variable as input parameter of train_test_split 
# Split data - train_test_split() - 67% training, 33% testing
x_train, x_test, y_train, y_test = train_test_split(dataset["text"],y,test_size=0.33, random_state=53)
```

```python
# create vectors - convert text to numerical vectors 
# Vectorization - CountVectorizer() - Converts text → numerical word-count vectors
count_vectorizer = CountVectorizer(stop_words='english')
count_train = count_vectorizer.fit_transform(x_train)
print(count_train)
count_test = count_vectorizer.transform(x_test)
#count_test
```

```python
len(count_vectorizer.get_feature_names_out())
```

```python
print(count_train.toarray())
```

```python
# validate by evalution metrics - MultinomialNB
# We cannot directly give text as input to the models but after convert to vector then given as input to model
# Train model - MultinomialNB() - Multinomial Naive Bayes - Learns patterns from the vectors
clf = MultinomialNB() 
clf.fit(count_train,y_train)

# Predict - clf.predict(count_test) - Predicts FAKE or REAL
pred = clf.predict(count_test)

# Evaluate - accuracy_score() - Measures prediction accuracy
# War between dependent - output and prediction
score = metrics.accuracy_score(y_test, pred)
print("accuracy: %0.3f" % score)

cm = metrics.confusion_matrix(y_test, pred, labels = ['FAKE','REAL'])
```

```python
# Detailed evaluation - classification_report() - Shows precision, recall, F1-score
from sklearn.metrics import classification_report
report = classification_report(y_test, pred)
```
